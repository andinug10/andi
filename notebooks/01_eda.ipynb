```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load data
df = pd.read_csv('../dataset/diabetes.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
# 2. Audit biologically impossible zeros
zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
zero_counts = {c: (df[c] == 0).sum() for c in zero_cols}
print('Zeros count:', zero_counts)

# Convert zeros to NaN
df_clean = df.copy()
for c in zero_cols:
    df_clean[c] = df_clean[c].replace(0, np.nan)
```

```python
# 3. Target Distribution
outcome_counts = df_clean['Outcome'].value_counts()
print(outcome_counts)
print(f'Positive class percentage: {outcome_counts[1] / len(df_clean):.2%}')
```

```python
# 4. Correlation Heatmap
corr = df_clean.corr()
print('Top correlations with Outcome:')
print(corr['Outcome'].sort_values(ascending=False))
```
